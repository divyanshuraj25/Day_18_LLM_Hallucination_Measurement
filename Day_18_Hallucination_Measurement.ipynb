```python
def evaluate_answer(ground_truth, answer):
    gt = str(ground_truth).lower().strip()
    ans = str(answer).lower().strip()

    # Error response
    if ans.startswith("error:"):
        return "Hallucinated"

    # Exact or direct match
    if gt in ans or ans in gt:
        return "Correct"

    # Important keywords for partial matches
    gt_words = set(gt.replace(",", "").split())
    ans_words = set(ans.replace(",", "").split())

    overlap = len(gt_words & ans_words) / max(len(gt_words), 1)

    if overlap >= 0.3:
        return "Partially Correct"

    return "Hallucinated"


df["evaluation"] = df.apply(
    lambda row: evaluate_answer(
        row["ground_truth"],
        row["llm_answer"]
    ),
    axis=1
)

print("Evaluation completed!")

df[[
    "id",
    "question",
    "ground_truth",
    "llm_answer",
    "evaluation"
]]
```

```python
# Create a simple trusted knowledge base

knowledge_base = []

for _, row in df.iterrows():
    knowledge_base.append({
        "id": row["id"],
        "domain": row["domain"],
        "fact": row["ground_truth"]
    })

knowledge_df = pd.DataFrame(knowledge_base)

print("Knowledge base created!")
print("Total facts:", len(knowledge_df))

knowledge_df.head(10)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Convert knowledge-base facts into searchable text
knowledge_texts = knowledge_df["fact"].astype(str).tolist()

vectorizer = TfidfVectorizer()
knowledge_vectors = vectorizer.fit_transform(knowledge_texts)


def retrieve_context(question, top_k=1):
    """
    Retrieve the most relevant fact from the knowledge base.
    """

    query_vector = vectorizer.transform([question])

    similarities = cosine_similarity(
        query_vector,
        knowledge_vectors
    )[0]

    best_indices = similarities.argsort()[-top_k:][::-1]

    results = []

    for idx in best_indices:
        results.append({
            "fact": knowledge_df.iloc[idx]["fact"],
            "similarity": similarities[idx],
            "domain": knowledge_df.iloc[idx]["domain"]
        })

    return results


# Test retrieval
test_question = "What is the chemical formula of water?"

results = retrieve_context(test_question)

print("Question:", test_question)
print("\nRetrieved context:")

for result in results:
    print("Fact:", result["fact"])
    print("Similarity:", round(result["similarity"], 4))
```

```python
def normalize_text(text):
    return (
        str(text)
        .lower()
        .replace(".", "")
        .replace(",", "")
        .replace("°", "")
        .strip()
    )


def local_rag_evaluation(ground_truth, answer):
    answer = normalize_text(answer)
    truth = normalize_text(ground_truth)

    # API errors are not hallucinations
    if answer == "api error" or answer.startswith("error:"):
        return "API Error"

    # Exact match
    if truth == answer:
        return "Correct"

    # Important equivalent answers
    equivalent_answers = {
        "seven": ["7"],
        "7": ["seven"],
        "ancient egyptians": ["ancient egypt", "egyptians"],
        "japanese yen": ["japanese yen"],
        "the roman empire": ["roman empire"],
        "100 degrees celsius": ["100 c", "100 °c"],
    }

    if answer in equivalent_answers.get(truth, []):
        return "Correct"

    # Ground truth contained in answer
    if truth in answer:
        return "Correct"

    return "Partially Correct"


df["rag_evaluation"] = df.apply(
    lambda row: local_rag_evaluation(
        row["ground_truth"],
        row["rag_answer"]
    ),
    axis=1
)

print("RAG evaluation completed!")

print(
    df["rag_evaluation"].value_counts()
)

df[[
    "id",
    "ground_truth",
    "rag_answer",
    "rag_evaluation"
]]
```

```python
import re


def get_words(text):
    text = str(text).lower()
    return set(re.findall(r"\b[a-z0-9]+\b", text))


def jaccard_similarity(text1, text2):
    words1 = get_words(text1)
    words2 = get_words(text2)

    if not words1 or not words2:
        return 0.0

    intersection = words1 & words2
    union = words1 | words2

    return len(intersection) / len(union)


def citation_signal(answer):
    answer = str(answer).lower()

    citation_patterns = [
        "source:",
        "sources:",
        "reference:",
        "references:",
        "http://",
        "https://",
        "[1]",
        "[2]"
    ]

    return any(pattern in answer for pattern in citation_patterns)


def contradiction_signal(ground_truth, answer):
    """
    Simple contradiction check for this factual dataset.
    Returns True when an obvious conflicting answer is detected.
    """

    gt = str(ground_truth).lower()
    ans = str(answer).lower()

    # Numeric contradiction
    numbers_gt = set(re.findall(r"\b\d+\b", gt))
    numbers_ans = set(re.findall(r"\b\d+\b", ans))

    if numbers_gt and numbers_ans:
        if numbers_gt.isdisjoint(numbers_ans):
            return True

    return False


def hallucination_detector(question, ground_truth, answer, retrieved_context):
    answer = str(answer)

    if answer.startswith("ERROR:") or answer == "API Error":
        return {
            "unsupported_claim": None,
            "jaccard_overlap": 0.0,
            "contradiction": None,
            "citation_present": False,
            "status": "API Error"
        }

    similarity = jaccard_similarity(
        answer,
        retrieved_context
    )

    contradiction = contradiction_signal(
        ground_truth,
        answer
    )

    citation = citation_signal(answer)

    unsupported = similarity < 0.10

    if contradiction or unsupported:
        status = "Potential Hallucination"
    else:
        status = "Supported"

    return {
        "unsupported_claim": unsupported,
        "jaccard_overlap": round(similarity, 4),
        "contradiction": contradiction,
        "citation_present": citation,
        "status": status
    }


print("4-signal hallucination detector created!")
```

```python
detector_results = []

for _, row in df.iterrows():
    retrieved = retrieve_context(row["question"], top_k=1)[0]

    result = hallucination_detector(
        question=row["question"],
        ground_truth=row["ground_truth"],
        answer=row["llm_answer"],
        retrieved_context=retrieved["document"]
    )

    detector_results.append(result)

detector_df = pd.DataFrame(detector_results)

df = pd.concat(
    [df.reset_index(drop=True), detector_df.reset_index(drop=True)],
    axis=1
)

print("Detector analysis completed!")
```

```python
import re

def normalize_answer(text):
    text = str(text).lower().strip()
    text = re.sub(r"[^\w\s]", "", text)

    replacements = {
        "seven": "7",
        "ancient egyptians": "ancient egypt",
        "the roman empire": "roman empire",
        "japanese yen": "japanese yen"
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    return text


def improved_evaluation(ground_truth, answer):
    if str(answer).startswith("ERROR:") or str(answer) == "API Error":
        return "API Error"

    truth = normalize_answer(ground_truth)
    ans = normalize_answer(answer)

    if truth == ans or truth in ans or ans in truth:
        return "Correct"

    return "Partially Correct"


df["evaluation"] = df.apply(
    lambda row: improved_evaluation(
        row["ground_truth"],
        row["llm_answer"]
    ),
    axis=1
)

print(df["evaluation"].value_counts())
```
